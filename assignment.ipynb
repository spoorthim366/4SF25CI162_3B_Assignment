```python
import pandas as pd
import numpy as np

# Create a DataFrame with 3 columns and 5 rows of random integers
df = pd.DataFrame(np.random.randint(1, 100, size=(5, 3)),
                  columns=['A', 'B', 'C'])

# Add a new column that is the product of the first two columns
df['Product'] = df['A'] * df['B']

# Display the DataFrame
print(df)
```

```python
import pandas as pd
import numpy as np

# Create a DataFrame with 3 columns and 4 rows
df = pd.DataFrame(np.random.randint(1, 100, size=(4, 3)),
                  columns=['A', 'B', 'C'])

print("Original DataFrame:")
print(df)

# Calculate row-wise sum
print("\nRow-wise Sum:")
print(df.sum(axis=1))

# Calculate column-wise sum
print("\nColumn-wise Sum:")
print(df.sum(axis=0))
```

```python
import pandas as pd
import numpy as np

# Create a DataFrame with 3 columns and 5 rows
df = pd.DataFrame(np.random.randint(1, 100, size=(5, 3)),
                  columns=['A', 'B', 'C'])

# Introduce some NaN values
df.iloc[1, 0] = np.nan
df.iloc[2, 1] = np.nan
df.iloc[3, 2] = np.nan

print("Original DataFrame:")
print(df)

# Fill NaN values with the mean of respective columns
df = df.fillna(df.mean())

print("\nDataFrame after filling NaN values:")
print(df)
```

```python
import pandas as pd
import numpy as np

# Create a DataFrame with Category and Value columns
df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], size=10),
    'Value': np.random.randint(1, 100, size=10)
})

print("Original DataFrame:")
print(df)

# Group by Category and calculate sum and mean
result = df.groupby('Category')['Value'].agg(['sum', 'mean'])

print("\nData Aggregation:")
print(result)
```

```python
import pandas as pd
import numpy as np

# Create a DataFrame with random data
df = pd.DataFrame({
    'Product': np.random.choice(['Laptop', 'Mobile', 'Tablet'], size=10),
    'Category': np.random.choice(['Electronics', 'Furniture', 'Clothing'], size=10),
    'Sales': np.random.randint(100, 1000, size=10)
})

print("Original DataFrame:")
print(df)

# Group by Category and calculate total sales
result = df.groupby('Category')['Sales'].sum()

print("\nTotal Sales for Each Category:")
print(result)
```

```python
import pandas as pd

# Create first DataFrame
df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Amit', 'Rahul', 'Priya', 'Sneha']
})

# Create second DataFrame
df2 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Marks': [85, 90, 78, 95]
})

print("First DataFrame:")
print(df1)

print("\nSecond DataFrame:")
print(df2)

# Merge DataFrames using common column
merged_df = pd.merge(df1, df2, on='ID')

print("\nMerged DataFrame:")
print(merged_df)
```

```python
import pandas as pd

# Create a MultiIndex
index = pd.MultiIndex.from_tuples(
    [('A', 1), ('A', 2), ('B', 1), ('B', 2)],
    names=['Category', 'ID']
)

# Create a DataFrame with MultiIndex
df = pd.DataFrame({
    'Name': ['Amit', 'Rahul', 'Priya', 'Sneha'],
    'Marks': [85, 90, 78, 95]
}, index=index)

print("Original MultiIndex DataFrame:")
print(df)

# Indexing: Select category A
print("\nIndexing - Category A:")
print(df.loc['A'])

# Access a specific row
print("\nAccessing a specific row:")
print(df.loc[('A', 1)])

# Slicing: Select rows from A to B
print("\nSlicing:")
print(df.loc[('A', 1):('B', 1)])
```

```python
import pandas as pd
import numpy as np

# Create MultiIndex
index = pd.MultiIndex.from_tuples(
    [('A', 'X'), ('A', 'Y'), ('B', 'X'), ('B', 'Y'), ('C', 'X')],
    names=['Category', 'SubCategory']
)

# Create DataFrame with random values
df = pd.DataFrame({
    'Value': np.random.randint(1, 100, size=5)
}, index=index)

print("Original DataFrame:")
print(df)

# Calculate sum for each Category and SubCategory
result = df.groupby(['Category', 'SubCategory']).sum()

print("\nSum of Values:")
print(result)
```

```python
import pandas as pd
import numpy as np

# Create a DataFrame with 3 columns and 5 rows
df = pd.DataFrame(np.random.randint(1, 100, size=(5, 3)),
                  columns=['A', 'B', 'C'])

print("Original DataFrame:")
print(df)

# Apply a function to double all values
result = df.apply(lambda x: x * 2)

print("\nDataFrame after doubling the values:")
print(result)
```

```python
import pandas as pd
import numpy as np

# Create a DataFrame with 3 columns and 6 rows
df = pd.DataFrame(np.random.randint(1, 100, size=(6, 3)),
                  columns=['A', 'B', 'C'])

print("Original DataFrame:")
print(df)

# Apply lambda function to create a new column
df['Sum'] = df.apply(lambda x: x['A'] + x['B'] + x['C'], axis=1)

print("\nDataFrame after adding Sum column:")
print(df)
```

```python
import pandas as pd
import numpy as np

# Create a Series with 5 random text strings
s = pd.Series(np.random.choice(['apple', 'banana', 'mango', 'orange', 'grapes'], size=5))

print("Original Series:")
print(s)

# Convert all strings to uppercase
result = s.str.upper()

print("\nSeries after converting to uppercase:")
print(result)
```
